## Exercise 1

```
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)

plt.figure(figsize=(6,4))
plt.plot(x, 2*x + 1, '-', label='y = 2x + 1')
plt.plot(x, 2*x + 2, '--', label='y = 2x + 2')
plt.plot(x, 2*x + 3, ':', label='y = 2x + 3')

plt.title('Lines y = 2x + c')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

## Exercise 3

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv('weight-height.csv')

length = data['Height'].to_numpy()
weight = data['Weight'].to_numpy()

# Convert units
length_cm = length * 2.54
weight_kg = weight * 0.45359237

# Means
mean_length = np.mean(length_cm)
mean_weight = np.mean(weight_kg)

print('Mean length (cm):', mean_length)
print('Mean weight (kg):', mean_weight)

plt.hist(length_cm, bins=20)
plt.title('Histogram of Lengths')
plt.xlabel('Length (cm)')
plt.ylabel('Frequency')
plt.show()
```

## Exercise 4

```
import numpy as np

A = np.array([[1,2,3],
              [0,1,4],
              [5,6,0]])

A_inv = np.linalg.inv(A)

print('Inverse matrix:')
print(A_inv)

print('\nA * A_inv =')
print(A @ A_inv)

print('\nA_inv * A =')
print(A_inv @ A)
```
